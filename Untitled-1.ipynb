```python
import requests
import pandas as pd
```

```python
#headers from web page
headers = {
    "authority":"www.vivino.com",
    "accept-language": "en-US,en;q=0.9",
    "user-agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/134.0.0.0 Safari/537.36 Edg/134.0.0.0",
    "content-type" : "application/json",
    "accept" : "application/json"
} #>> usually needed

#parameters
params = {"per_page":"20", "year":"2022"}
```

```python
response = requests.get("https://www.vivino.com/api/wines/6139455/reviews?per_page=4&year=N.V.&language=en", params=params, headers=headers)
response.status_code
response.json()
len(response.json()["reviews"])
```
